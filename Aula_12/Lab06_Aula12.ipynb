```python
import numpy as np
from pymoo.core.problem import ElementwiseProblem
from pymoo.algorithms.moo.nsga2 import NSGA2
from pymoo.optimize import minimize

class EVBatteryProblem(ElementwiseProblem):
    def __init__(self):
        # TODO 1: Configurar restrição de desigualdade (n_ieq_constr=1)
        super().__init__(n_var=2, n_obj=2, n_ieq_constr=1, xl=np.array([10.0, 15.0]), xu=np.array([150.0, 45.0]))

    def _evaluate(self, x, out, *args, **kwargs):
        potencia, temp = x[0], x[1]

        f1_tempo_recarga = 3000.0 / potencia
        f2_desgaste_bateria = (potencia ** 1.5) * (temp / 20.0)

        # TODO 2: Restrição de segurança G(x) <= 0
        g1 = f2_desgaste_bateria - 1500.0

        out["F"] = [f1_tempo_recarga, f2_desgaste_bateria]

        # TODO 3: Passar a restrição para a saída
        out["G"] = [g1]

# TODO 4: Execute a otimização com o NSGA-II e imprima as soluções viáveis
res6 = minimize(EVBatteryProblem(), NSGA2(pop_size=30), termination=('n_gen', 40), seed=6)

print("--- Exercício 6: Soluções Viáveis (Tempo, Desgaste) ---")
for f in res6.F:
    print(f"Tempo de Recarga: {f[0]:.2f} | Desgaste: {f[1]:.2f}")
```
