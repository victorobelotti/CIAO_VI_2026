```python
import random

# 1. Dados simulados do problema (ex: extraídos do ERP)
# Cada pedido tem um ID, Peso (kg) e Prioridade (1 a 10)
pedidos = [
    {"id": "PED001", "peso": 1500, "prioridade": 8},
    {"id": "PED002", "peso": 800,  "prioridade": 5},
    {"id": "PED003", "peso": 2100, "prioridade": 10},
    {"id": "PED004", "peso": 600,  "prioridade": 4},
    {"id": "PED005", "peso": 1200, "prioridade": 7},
    {"id": "PED006", "peso": 400,  "prioridade": 6}
]

CAPACIDADE_MAXIMA = 3500 # Limite do veículo em kg

def gerar_solucao_aleatoria(n):
    """Gera um vetor binário aleatório indicando quais pedidos vão (1) e quais ficam (0)."""
    return [random.choice([0, 1]) for _ in range(n)]

def calcular_funcao_objetivo(solucao, lista_pedidos):
    """Calcula a soma das prioridades dos pedidos que receberam '1'."""
    prioridade_total = 0
    for i in range(len(solucao)):
        if solucao[i] == 1:
            prioridade_total += lista_pedidos[i]["prioridade"]
    return prioridade_total

def verificar_restricoes(solucao, lista_pedidos, limite_peso):
    """Verifica se o peso dos itens escolhidos estoura a capacidade do caminhão."""
    peso_total = 0
    for i in range(len(solucao)):
        if solucao[i] == 1:
            peso_total += lista_pedidos[i]["peso"]

    # É factível se o peso total for menor ou igual à capacidade
    solucao_valida = peso_total <= limite_peso
    return solucao_valida, peso_total

# ==========================================
# TESTANDO A MODELAGEM
# ==========================================

n_pedidos = len(pedidos)

# Passo A: Gerar uma solução aleatória
solucao_candidata = gerar_solucao_aleatoria(n_pedidos)

# Passo B: Calcular a função objetivo
objetivo = calcular_funcao_objetivo(solucao_candidata, pedidos)

# Passo C: Verificar as restrições
valida, peso_calculado = verificar_restricoes(solucao_candidata, pedidos, CAPACIDADE_MAXIMA)

# Exibindo os resultados
print(f"Vetor Solução Aleatória: {solucao_candidata}")
print(f"Soma das Prioridades (Função Objetivo): {objetivo}")
print(f"Peso Total da Carga: {peso_calculado} kg (Limite: {CAPACIDADE_MAXIMA} kg)")
print(f"A solução respeita a restrição de peso? {'Sim (Solução Factivel)' if valida else 'Não (Solução Inválida)'}")
```
