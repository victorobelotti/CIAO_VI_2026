```python
import numpy as np

# Configurações do Problema
TAREFAS = np.array([12, 35, 40, 8, 15, 22, 19, 45, 60, 31, 14, 28, 50, 18, 25, 33, 42, 10, 5, 29])
NUM_TAREFAS = len(TAREFAS)
NUM_SERVIDORES = 4

# Parâmetros do Algoritmo Genético
TAM_POP = 50
GERACOES = 100
TAXA_MUTACAO = 0.1

def calcular_makespan(individuo):
    """
    Calcula o tempo total de cada servidor e retorna o maior deles (Makespan).
    """
    cargas = np.zeros(NUM_SERVIDORES)
    for i_tarefa, i_servidor in enumerate(individuo):
        cargas[i_servidor] += TAREFAS[i_tarefa]
    return np.max(cargas)

# 1. Inicialização da População
populacao = [np.random.randint(0, NUM_SERVIDORES, size=NUM_TAREFAS) for _ in range(TAM_POP)]

for g in range(GERACOES):
    custos = [calcular_makespan(ind) for ind in populacao]
    melhor_idx = np.argmin(custos)

    novos = []
    # Elitismo: preserva o melhor indivíduo
    novos.append(populacao[melhor_idx].copy())

    while len(novos) < TAM_POP:
        # Seleção por Torneio (tamanho 2)
        i1, i2 = np.random.choice(TAM_POP, 2, replace=False)
        pai1 = populacao[i1] if custos[i1] < custos[i2] else populacao[i2]

        i3, i4 = np.random.choice(TAM_POP, 2, replace=False)
        pai2 = populacao[i3] if custos[i3] < custos[i4] else populacao[i4]

        # Cruzamento (Crossover de ponto único)
        ponto_corte = np.random.randint(1, NUM_TAREFAS)
        filho = np.concatenate([pai1[:ponto_corte], pai2[ponto_corte:]])

        # Mutação: altera o servidor de uma tarefa aleatória
        if np.random.rand() < TAXA_MUTACAO:
            pos_mutacao = np.random.randint(0, NUM_TAREFAS)
            filho[pos_mutacao] = np.random.randint(0, NUM_SERVIDORES)

        novos.append(filho)

    populacao = novos

# Resultado Final
custos_finais = [calcular_makespan(ind) for ind in populacao]
melhor_idx = np.argmin(custos_finais)
melhor_solucao = populacao[melhor_idx]
menor_makespan = custos_finais[melhor_idx]

# Exibição dos detalhes da alocação
cargas_finais = np.zeros(NUM_SERVIDORES)
for i_tarefa, i_servidor in enumerate(melhor_solucao):
    cargas_finais[i_servidor] += TAREFAS[i_tarefa]

print(f"[Exercício 3] Menor Makespan Encontrado: {menor_makespan}s")
print("\nDistribuição de Carga por Servidor:")
for s in range(NUM_SERVIDORES):
    print(f"  Servidor {s}: {cargas_finais[s]}s")
```
